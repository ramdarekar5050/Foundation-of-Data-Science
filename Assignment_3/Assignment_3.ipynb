```python
data ={
    'Student':['s1', 's2', 's3', 's4', 's5', 's6', 's7', 's8', 's9', 's10'],
    'Marks' : [12, 15, 12, 18, 19, 12, 17, 20, 18, 15]
    }
```

```python
# Create DataFrame and save to CSV
import pandas as pd
df = pd.DataFrame(data)
df.to_csv('data.csv', index=False)

```

```python
# Read CSV file
df = pd.read_csv('data.csv')
print(df)
```

```python
marks = df['Marks'].tolist()
```

```python
# Mean calculation function
def calculate_mean(data):
    return sum(data) / len(data)
print("Mean:", calculate_mean(marks))
```

```python
# Median calculation function
def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2
    if n % 2 == 0:
        median = (sorted_data[mid - 1] + sorted_data[mid]) / 2
    else:
        median = sorted_data[mid]
    return median
print("Median:", calculate_median(marks))
```

```python
# Mode calculation function
def calculate_mode(data):
    freq = {}
    for val in data:
        freq[val] = freq.get(val, 0) + 1
    max_freq = max(freq.values())
    mode = [key for key, value in freq.items() if value == max_freq]
    return mode
print("Mode:", calculate_mode(marks))   
```
