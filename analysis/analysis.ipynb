```python
#Name: Ram Darekar
#Roll NO: 114
#Division: B


import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = {
    'Product': ['Laptop A', 'Phone X', 'Tablet Pro', 'Monitor HD', 'Keyboard Z', 'Mouse Y'],     #Qualitative data
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories', 'Accessories', 'Accessories'], # Qualitative data
    'Price ($)': [1200, 800, 600, 300, 50, 40],      # Quantitative data
    'Stock Quantity': [25, 50, 30, 40, 100, 150],     # Quantitative data
    'Supplier': ['TechCorp', 'MobileInc', 'GadgetWorld', 'DisplayMart', 'KeyMasters', 'ClickStore'],  # Qualitative data
    'Rating': [4.5, 4.2, 4.0, 4.3, 4.1, 4.6], # Quantitative data
    'Warranty (Years)': [2, 1, 2, 3, 1, 1]
}

df = pd.DataFrame(data)
df.head()
```

```python
# --- QUALITATIVE ANALYSIS ---
print("Qualitative Analysis:")

# Frequency of categories
print("\nCategory Frequency:")
print(df['Category'].value_counts())

# Frequency of suppliers
print("\nSupplier Frequency:")
print(df['Supplier'].value_counts())

# Bar chart for Category distribution
df['Category'].value_counts().plot(kind='bar', color='skyblue')
plt.title("Product Category Distribution")
plt.xlabel("Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python


# --- QUANTITATIVE ANALYSIS ---
print("\nQuantitative Analysis:")
print(df[['Price ($)', 'Stock Quantity', 'Rating', 'Warranty (Years)']].describe())

# Histogram for Price
df['Price ($)'].plot(kind='hist', bins=5, color='orange', edgecolor='black')
plt.title("Price Distribution")
plt.xlabel("Price ($)")
plt.ylabel("Frequency")
plt.show()

# Histogram for Stock Quantity
df['Stock Quantity'].plot(kind='hist', bins=5, color='purple', edgecolor='black')
plt.title("Stock Quantity Distribution")
plt.xlabel("Stock Quantity")
plt.ylabel("Frequency")
plt.show()

# Histogram for Rating
df['Rating'].plot(kind='hist', bins=5, color='green', edgecolor='black')
plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()


```

```python
# Scatter plot: Price vs Rating
plt.scatter(df['Price ($)'], df['Rating'], color='blue', edgecolor='black')
plt.title("Price vs. Customer Rating")
plt.xlabel("Price ($)")
plt.ylabel("Rating")
plt.grid(True)
plt.show()
```

```python
# Scatter plot: Warranty vs Stock Quantity
plt.scatter(df['Warranty (Years)'], df['Stock Quantity'], color='darkgreen', edgecolor='black')
plt.title("Warranty vs. Stock Quantity")
plt.xlabel("Warranty (Years)")
plt.ylabel("Stock Quantity")
plt.grid(True)
plt.show()
```

```python
# Histogram for Warranty
df['Warranty (Years)'].plot(kind='hist', bins=4, color='teal', edgecolor='black')
plt.title("Warranty Period Distribution")
plt.xlabel("Warranty (Years)")
plt.ylabel("Frequency")
plt.show()
```
