```python
data ={
    'Student':['s1', 's2', 's3', 's4', 's5', 's6', 's7', 's8', 's9', 's10'],
    'Marks' : [12, 15, 12, 18, 19, 12, 17, 20, 18, 15]
    }
```

```python
# Create DataFrame and save to CSV
import pandas as pd
df = pd.DataFrame(data)
df.to_csv('data.csv', index=False)

```

```python
# Read CSV file
df = pd.read_csv('data.csv')
print(df)
```

```python
marks = df['Marks'].tolist()
```

```python
# Range
def calculate_range(marks):
    return max(marks) - min(marks)
print("Range:", calculate_range(marks))
```

```python
# Variance
def calculate_variance(marks):
    mean = sum(marks) / len(marks)
    squared_diffs = [(x - mean) ** 2 for x in marks]
    variance = sum(squared_diffs) / (len(marks) - 1)  # sample variance
    return variance

print("Variance:", calculate_variance(marks))
```

```python
# Standard Deviation
def calculate_std_dev(marks):
    variance = calculate_variance(marks)
    return variance ** 0.5

print("Standard Deviation:", calculate_std_dev(marks))
```

```python
import statistics as stats



# Range
range_value = max(marks) - min(marks)

# Variance
variance_value = stats.variance(marks)

# Standard Deviation
std_dev_value = stats.stdev(marks)

print("Using Inbuilt Functions:")
print("Range:", range_value)
print("Variance:", variance_value)
print("Standard Deviation:", std_dev_value)
```
