```python
# Import libraries

import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd 
```

```python
# Graphs Using Matplotlib
```

```python

# Sample data
Students = ['Ram', 'Pajju', 'Om', 'Jay','Don']
marks = [5, 7, 3, 8, 4]

# Plotting the bar graph
plt.bar(Students, marks,color='skyblue',edgecolor='black')

# Adding title and labels
plt.title("Simple Bar Graph")
plt.xlabel("Students")
plt.ylabel("marks")

# Show the plot
plt.show()
```

```python
# Sample data
data = [12, 15, 13, 15, 18, 21, 22, 19, 24, 30, 22, 25, 20, 18, 17, 16, 20, 23]

# Plotting the histogram
plt.hist(data, bins=5,color='orange',edgecolor='black') 

# Adding title and labels
plt.title("Simple Histogram")
plt.xlabel("Value Ranges")
plt.ylabel("Frequency")
```

```python
# Sample data
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

# Plotting the graph
plt.plot(x, y,color='blue', marker='o', linestyle='--', label='Line Plot')

# Adding title and labels
plt.title("Simple Line Graph")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()  # Show legend
plt.grid(True)  # Add grid
plt.show()
```

```python

# Sample data
x = [5, 7, 8, 7, 2, 17, 2, 9, 4, 11]
y = [99, 86, 87, 88, 100, 86, 103, 87, 94, 78]


# Adding title and labels
plt.scatter(x, y, color='red', marker='*')
plt.title("Scatter Plot Example")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python
# Graphs Using Seaborn
```

```python
# Sample dataset
data = {'Student': ['Ram', 'Pajju', 'Om', 'Jay'],
        'Marks': [90, 75, 80, 85]}

df = pd.DataFrame(data)

# Bar plot using seaborn
sns.barplot(x='Student', y='Marks', data=df, palette='Blues_d')
plt.title("Seaborn Bar Plot")
plt.show()
```

```python
# Sample data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
temp = [55, 60, 55, 63, 69]

sns.lineplot(x=months, y=temp, marker='o', color='purple')
plt.title("Temperature Over a Week")
plt.xlabel("Months")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
# Sample data
marks = [90, 85, 96, 90, 95, 80, 87,87,96, 85, 95, 80, 80, 95]

# Create histogram
sns.histplot(marks, bins=5, color='skyblue', kde=True)  # kde=True adds smooth curve

# Add title and labels
plt.title("Marks Distribution (Seaborn Histogram)")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python
# Sample data
x = [3, 6, 9, 4, 1, 12, 5, 8]
y = [75, 80, 79, 82, 90, 77, 95, 83]

# Create scatter plot
sns.scatterplot(x=x, y=y, color='red', s=100)  

# Add title and labels
plt.title("Scatter Plot Example (Seaborn)")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
```

```python


# Sample data (correlation-like table)
data = {
    'Math': [65, 78, 84, 91],
    'Physics': [72, 69, 85, 88],
    'Chemistry': [81, 76, 89, 93],
    'Biology': [70, 82, 77, 85]
}

# Convert to DataFrame
df = pd.DataFrame(data, index=['Student A', 'Student B', 'Student C', 'Student D'])

# Create heatmap
sns.heatmap(df, annot=True, cmap='YlGnBu', linewidths=0.5)

# Add title
plt.title("Students Marks Heatmap")
plt.show()
```
