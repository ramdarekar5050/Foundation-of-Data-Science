```python
data={
    'Name':['Ram','Abhi','Pajju','kuldeep'],
     'Age':[21,22,19,18],
     'cgpa':[8.5,9.0,7.5,8.0],
     'marks':[85,90,75,80]  
    
     }  


df=pd.DataFrame(data)
print("\nDataframe\n")
print(df)
print("\nAccessing elements by index:")    
print(df['Name'])
```
