```python
data = {
    'Name': ['Ram', 'Prajwal', 'Om', 'Jay', 'Sham'],
    'Class': ['A', 'A', 'B', 'B', 'A'],
    'Marks': [70, 75, 78, 90, 99]
}

df3 = pd.DataFrame(data)
print(df3.groupby('Class')['Marks'].min())
```

```python
data = {
    'Department': ['IT','IT','EC','CS','Mech','Mech','IT','CS'],
    'Student': ['Ram','Riya','Prajwal','Meena','Aniket','Neha','Rahul','Jay'],
    'Marks': [90,88,85,70,99,90,75,72],
    'Age': [19,21,22,19,21,20,22,23]
}

df = pd.DataFrame(data)
print(df)
```

```python
df.groupby('Department')['Age'].max()       # max age in each department
df.groupby('Department')['Marks'].mean()    # average marks in each department
```

```python
df.groupby('Department')['Marks'].count()   # count of students in each department
```

```python
df.groupby('Department')['Marks'].agg(['mean','max','min','count'])         # aggregate functions
```

```python
df.groupby(['Department','Age'])['Marks'].max()   # multi-level grouping
```

```python
df['Marks'].plot(kind='hist')     # histogram
```

```python
df.plot(x='Department', y='Marks', kind='scatter')  # scatter plot
```

```python
df['Marks'].plot(kind='box')      # boxplot
```

```python
df4.isnull().sum()              # check missing values
```

```python
df4.fillna(0, inplace=True)     # fill NaN with 0
df4.fillna(df.min())           # fill NaN with mean
df4.dropna()                    # remove rows with NaN
df4.duplicated().sum()          # check duplicates
df4.drop_duplicates()  # remove duplicates
```

```python
df4['WorkHours'] = df4['WorkHours'].fillna(df4['WorkHours'].mean())   # fill NaN with mean of column
```

```python
df4.isnull().sum()              # check missing values
```

```python
df4.groupby('Department')['Salary'].mean()    # average salary in each department
```

```python
df4.groupby(['Department', 'Gender'])['Salary'].min()   # average salary by Department and Gender
```

```python
df4.groupby('Department')['Salary'].agg(['count', 'mean', 'sum', 'min', 'max', 'std', 'median'])   # aggregate functions
```

```python
df4['Salary'].plot(kind='hist')     # histogram
```

```python
df4.plot.scatter(x='Age', y='Salary', alpha=0.5, title='Age vs Salary')     # scatter plot
```
